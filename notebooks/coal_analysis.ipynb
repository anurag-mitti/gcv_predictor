### 3. Interactive SHAP Explainability
This step natively visualizes the interactive JS-based SHAP force plots right here in the notebook.

```python
import pandas as pd
import joblib
import shap

# Initialize Javascript rendering in Jupyter
shap.initjs()

# Load Model & Data
model = joblib.load('../models/ensemble_gcv.joblib')
df = pd.read_csv('../data/processed_coal.csv')
X = df[['Moisture', 'Volatile Matter', 'Standard Ash', 'Fixed Carbon', 'Fuel_Ratio']]

# Extract XGBoost component and run Explainer
xgb_model = model.named_estimators_['xgb']
explainer = shap.TreeExplainer(xgb_model)
shap_values = explainer(X)

# Display interactive force plot for the first sample
shap.plots.force(shap_values[0])
```
